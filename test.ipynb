```python
# debug_ic_try.py
import os
import pandas as pd
import numpy as np

FEATURE_DIR = "data/features"
WIDTH_CANDIDATES = [100,200,300,400,500,750,1000,1500,2000]
MIN_CREDIT_TO_WIDTH = 0.12   # match script defaults (adjust if you changed via CLI)
MIN_CREDIT_ABS = 20.0

def nearest_available_strike(strikes, target):
    if len(strikes)==0 or pd.isna(target): 
        return None
    idx = np.abs(strikes - target).argmin()
    return float(strikes[idx])

def find_option_price(df, strike, optype):
    sub = df[(df['strike'] == strike) & (df['option_type'].str.upper() == optype.upper())]
    if sub.empty:
        return None
    sub = sub.assign(_close_nonzero=(sub['close'].fillna(0) > 0).astype(int))
    sub = sub.sort_values(by=['_close_nonzero','close'], ascending=[False, True])
    val = sub['close'].iloc[0]
    if pd.isna(val) or float(val) == 0.0:
        return None
    return float(val)

def inspect_date(date_str):
    p = os.path.join(FEATURE_DIR, f"{date_str}.parquet")
    if not os.path.exists(p):
        print("Feature file missing:", p); return
    df_all = pd.read_parquet(p)
    print("Loaded features:", p)
    entry_dt = pd.to_datetime(date_str)

    # list expiries (sorted)
    exps = sorted(pd.to_datetime(df_all['expiry'].dropna().unique()))
    print("Expiries (first 12):", exps[:12])
    # pick nearest expiry (min dte)
    exps_after = [e for e in exps if e > entry_dt]
    if not exps_after:
        print("No future expiries")
        return
    expiry = exps_after[0]
    dte = (expiry - entry_dt).days
    print("Selected expiry:", expiry, "DTE:", dte)
    df = df_all[pd.to_datetime(df_all['expiry']) == expiry].copy()
    print("Rows in chain:", len(df))

    # regime + iv
    print("Underlying unique:", df['underlying'].dropna().unique())
    print("nifty_adx_14 (unique):", df['nifty_adx_14'].dropna().unique())
    print("nifty_bb_bw_20 (unique):", df['nifty_bb_bw_20'].dropna().unique())
    print("iv_pctile present:", 'iv_pctile_local' in df.columns and df['iv_pctile_local'].notna().any())
    if 'iv_pctile_local' in df.columns:
        print("iv_pctile sample:", df['iv_pctile_local'].dropna().unique()[:10])

    strikes = np.sort(df['strike'].dropna().unique())
    print("Strike count:", len(strikes))
    underlying = float(df['underlying'].dropna().unique()[0])
    print("Underlying:", underlying)
    mid_frac = (0.15 + 0.25)/2.0
    target_short_call = underlying * (1.0 + mid_frac)
    target_short_put  = underlying * (1.0 - mid_frac)
    print("Target short call (moneyness):", target_short_call)
    print("Target short put  (moneyness):", target_short_put)

    short_call = nearest_available_strike(strikes, target_short_call)
    short_put  = nearest_available_strike(strikes, target_short_put)
    print("Nearest short_call strike:", short_call)
    print("Nearest short_put strike :", short_put)

    print("\nNow trying widths and computing credits (width, long_call, long_put, p_sc, p_sp, p_lc, p_lp, credit, pass?)")
    for width in WIDTH_CANDIDATES:
        long_call_candidates = strikes[strikes >= short_call + width]
        long_put_candidates  = strikes[strikes <= short_put - width]
        if len(long_call_candidates)==0 or len(long_put_candidates)==0:
            print(f" width {width:5}: SKIP - no long strikes available")
            continue
        long_call = float(long_call_candidates[0])
        long_put  = float(long_put_candidates[-1])
        p_sc = find_option_price(df, short_call, 'CE')
        p_sp = find_option_price(df, short_put,  'PE')
        p_lc = find_option_price(df, long_call,  'CE')
        p_lp = find_option_price(df, long_put,   'PE')
        print(f" width {width:5}: long_call {long_call:.0f}, long_put {long_put:.0f}", end="   ")
        print(f" p_sc={p_sc} p_sp={p_sp} p_lc={p_lc} p_lp={p_lp}", end="   ")
        if None in (p_sc,p_sp,p_lc,p_lp):
            print(" -> MISSING PRICES")
            continue
        credit = (p_sc + p_sp) - (p_lc + p_lp)
        pass_flag = credit >= max(MIN_CREDIT_ABS, MIN_CREDIT_TO_WIDTH * width)
        print(f" -> credit={credit:.2f} pass={pass_flag}")

if __name__ == '__main__':
    # change dates list here as needed
    dates = ["2024-07-05"]
    for d in dates:
        print("\n\n==== DIAG:", d, "====")
        inspect_date(d)
```

```python
import requests
import json

url = "http://localhost:11434/api/chat"

payload = {
    "model": "llama3.2",
    "messages": [
        {"role": "user", "content": "Hello from debug test!"}
    ],
    "stream": False
}

print("Sending...")
r = requests.post(url, json=payload)

print("STATUS:", r.status_code)
print("RAW TEXT:")
print(r.text)

try:
    obj = json.loads(r.text)
    print("\nJSON Parsed OK:")
    print(json.dumps(obj, indent=2))
except Exception as e:
    print("\nJSON ERROR:", str(e))
```
