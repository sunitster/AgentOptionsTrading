```python
import yaml
from kiteconnect import KiteConnect

# -----------------------------------------------------
# 1. Load config/config.yaml (top-level keys only)
# -----------------------------------------------------
def load_config(path="config/config.yaml"):
    with open(path, "r") as f:
        return yaml.safe_load(f)

cfg = load_config()

# Read credentials from TOP LEVEL of YAML
api_key = cfg.get("api_key")
api_secret = cfg.get("api_secret")
access_token = cfg.get("access_token")

print("Loaded:")
print("  api_key       =", api_key)
print("  api_secret    =", "(hidden)" if api_secret else None)
print("  access_token  =", "(present)" if access_token else "(missing)")

# -----------------------------------------------------
# 2. Validate required fields
# -----------------------------------------------------
if not api_key or not api_secret:
    print("❌ ERROR: Missing api_key or api_secret in config/config.yaml")
    exit()

# -----------------------------------------------------
# 3. Initialize KiteConnect
# -----------------------------------------------------
kite = KiteConnect(api_key=api_key)

if not access_token:
    print("\n⚠️ No access_token found in config/config.yaml")
    print("👉 Generate it using:")
    print(f"https://kite.zerodha.com/connect/login?v=3&api_key={api_key}")
    exit()

kite.set_access_token(access_token)

# -----------------------------------------------------
# 4. Test LIVE authentication
# -----------------------------------------------------
print("\n🔍 Testing access token with LIVE Kite API...")

try:
    profile = kite.profile()
    print("✅ SUCCESS! Access token is valid.")
    print("Logged in as:", profile["user_id"])
except Exception as e:
    print("\n❌ INVALID ACCESS TOKEN")
    print("Reason:", e)
    print("\nTo generate a fresh token:")
    print(f"👉 Visit: https://kite.zerodha.com/connect/login?v=3&api_key={api_key}")
    print("After login → copy request_token → exchange it for access_token.")
```
